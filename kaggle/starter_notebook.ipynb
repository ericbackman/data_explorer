```python
import sys
sys.path.append('/kaggle/input/sports-dbs')   # so we can import the helper shipped with the data
import kaggle_sportsdb as sportsdb

sportsdb.databases()   # which DBs attached
```

```python
sportsdb.tables().groupby('database').size()   # tables per DB
```

```python
# Sample query — top single-game scorers. Validate against a known fact before trusting
# any derived number (the data_explorer discipline rule).
df = sportsdb.q(
    """
    SELECT p.player_name, g.pts, g.game_date
    FROM nba.player_game g
    JOIN nba.players p USING (player_id)
    ORDER BY g.pts DESC
    LIMIT 10
    """
)
df
```

```python
import matplotlib.pyplot as plt

ax = df[::-1].plot.barh(x='player_name', y='pts', legend=False, figsize=(8, 4))
ax.set_xlabel('Points')
ax.set_ylabel('')
ax.set_title('Top 10 single-game scoring performances')
plt.tight_layout()
```
